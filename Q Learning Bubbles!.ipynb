## Reinforcement Learning





Los tres principales paradigmas de Machine Learning son:

1) Supervised Learning

2) Unsupervised Learning

3) Reinforcement Learning


El objetivo principal de Reinforcement Learning es el de hacer que una máquina "entienda" su ambiente (environment) de manera tal que pueda aprender a tomar acciones que maximicen sus rewards acumuladas a lo largo del tiempo. 
Para esto, se debe hallar un balance en la proporción de acciones en la que la máquina explora el ambiente y aprende sobre experiencias "nuevas", y la proporción de acciones en las que la máquina utiliza lo ya aprendido sobre el ambiente en previas exploraciones.

Los algoritmos de Reinforcement Learning tienen como inputs, estados del sistema o ambiente en cuestión y rewards. Por otro lado, el output de un algoritmo de Reinforcement Learning es la acción que la máquina debe tomar para maximizar las rewards acumuladas a largo plazo.

El objetivo de un modelo de Reinforcement Learning será entonces el de hallar la política de acción óptima para maximizar las rewards acumuladas, es decir, hallar qué acción tomar en respuesta a cada estado para poder acumular rewards correctos. Por supuesto, habrá veces en las que la máquina deba sacrificar rewards inmediatas para obtener mayores rewards futuras.

En cuanto al ambiente, este representa la fuente de información de la máquina sobre los estados y los rewards. El ambiente, a su vez, define las "reglas de juego" (lo que es y no es posible hacer) que deberá respetar la máquina. Es entonces el ambiente el que define las acciones posibles de la máquina en determinado estado y es además el que define la transición de un estado en respuesta a la acción que la máquina tomó. 

Las mencionadas reglas de juego son inicialmente desconocidas por la máquina y se van aprendiendo a medida que el sistema explora y va obteniendo rewards.

El soporte matemático en el cual se basa el sistema de Reinforcement Learning, con el cual aprende qué decisiones tomar para maximizar sus rewards acumuladas está dado por el Markov Decision Process (MDP). A partir del MDP, el sistema de Reinforcement Learning en el tiempo t tomará una accion $a_t$ basándose en el estado $S_t$. Esta acción $a_t$ es elegida a partir de las posibles acciones pertenecientes al conjunto $A_t$ y estará basada una política de acción $P_t$ de forma tal que dada la reward obtenida para ese tiempo, llamada $r_t$, se maximicen las rewards acumuladas $R_t = \sum \limits _{i=0} ^{t} r_i$. Luego de tomada la accion $a_t$, se transicionará al estado $S_{t+1}$ y se repetirá el mismo ciclo. La política de acción se irá actualizando según el estado, la acción tomada, el reward recibido luego de tomar dicha acción y el próximo estado. 




## Q-learning

Q-learning es un tipo de Reinforcement Learning en el que la máquina intenta obtener una política óptima interactuando directamente con el ambiente, a través de prueba y error a lo largo de varios episodios. El agente no intenta, por lo tanto, aprender un modelo matemático o distribución de probabilidad. Es por estas razones que Q-learning se conoce como un "model free environment".

En Q-learning, la cantidad de estados posibles es finita y la cantidad de posibles acciones también lo es. Así, el agente de Q-learning deberá actualizar los llamados "Q-values" o valores-Q, que indica la calidad de una acción particular $a$ para un dado estado $s$. Podemos así escribir a la función $Q(a, s)$. El agente buscará elegir la acción con el mayor valor-Q para el dado estado a la hora de decidir qué acción tomar. 

Los valores-Q serán los estimadores actuales de la suma de rewards futuros si fuésemos a tomar la acción a en según ese estado. 

Para problemas de pequeña dimensión, los valores-Q se guardan y actualizan en una tabla de valores Q llamada "Q-table" o tabla-Q, que vendrá a representar la política de acción del agente. Cuando los problemas crecen en dimensión, se utilizan redes neuronales, cuyos coeficientes suplantan los valores de la tabla de una forma más comprimida y más fácil/rápido de actualizar. 

En relación a la actualización de los valores Q se introducen las "temporal differences", figuras que proveen un método de calcular cuanto el Q value de la acción tomada debería ser cambiado en función de lo que el agente ya aprendió sobre los Q-values de las distintas acciones del estado actual. Esto se representa en la siguiente ecuación:

$TD(s_t, a_t) = r_t + \gamma \cdot max_a Q(s_{t+1}, a) - Q(s_t, a_t)$

Donde $\gamma$ es el factor de descuento, donde $\gamma \epsilon [0,1]$, e indica la prioridad que tienen las rewards inmediatas en comparación con las lejanas en el tiempo. Cuanto más grande es $\gamma$, más se priorizarán las rewards inmediatas, por lo que $\gamma$ suele rondar valores de $0.9$.

Habiendo definido los "temporal differences" se utiliza la ecuación de Bellman que define que nuevo Q-value se tiene que utilizar para la acción tomada en el estado previo. La ecuación se presenta a continuación:

$Q^{new}(s_t, a_t) = Q^{old}(s_t, a_t) + \alpha.TD(s_t, a_t)$

Donde TD es la "temporal difference" calculada para el estado y acción dada y alfa es un learning rate definido utilizado para controlar la tasa de aprendizaje.

Para mejorar este aprendizaje a lo largo de los episodios se definen los conceptos de exploración y explotación. La exploración ocurre cuando el agente toma una acción aleatoria, permitiendo conocer posibles nuevos estados del ambiente en cuestión, mientras que explotación se refiere a cuando el agente toma la acción más óptima que considera.

Para delimitar las proporciones de exploración vs explotación define un parámetro llamado 'índice de exploración' que define la probabilidad que una acción dada de un episodio sea aleatoria o sea realizada teniendo en cuenta lo aprendido por la red hasta el momento.

# Modelo de Red utilizado y Consideraciones de entrenamiento

Conocido el juego y el método de aprendizaje que se va a utilizar (deep Q learning) se procedió a elegir el diseño de red neuronal que va a realizar el trabajo de agente. Para la elección del diseño se buscó en Internet bases y criterios tomados al momento de diseñar la misma. En particular el diseño propio se basó en un ejemplo provisto por [Keras](https://keras.io/examples/rl/deep_q_network_breakout) , donde se entrena a un modelo basado en el paper de DeepMind para aprender a jugar al juego de Atari Breakout.

A continuación se presenta el diseño de red planteado para este proyecto:

```python
def DQN(num_actions, activate):
    """https://keras.io/examples/rl/deep_q_network_breakout/"""

    input_lay = layers.Input(shape=(GAMEOVER_ROWS + 3, 2*GRID_COLS + 1, 5,))

    conv1 = layers.Conv2D(32, 4, strides=1, activation=activate)(input_lay)
    conv2 = layers.Conv2D(64, 2, strides=1, activation=activate)(conv1)

    flat = layers.Flatten()(conv2)
    dense1 = layers.Dense(512, activation=activate)(flat)
    dense2 = layers.Dense(512, activation=activate)(dense1)
    dense3 = layers.Dense(num_actions, activation='linear')(dense2)

    return keras.Model(inputs=input_lay, outputs=dense3)
```


Puede observarse que el diseño general del mismo es bastante similar al propuesto por Deepmind en el ejemplo de Keras con algunas modificaciones. La principal modificación fue el quitar una capa convolucional. Esta decisión fue tomada ya que en el caso de DeepMind se obtenían los pixeles directos de la pantalla en formato RGB como input al modelo, mientras que en nuestro caso se hace un previo separamiento en capas por colores (los objetos ya están detectados), por lo que se consideró que podía quitarse una de las capas convolucionales.

En relación a la activación de las capas internas de la red se probó con ambos tipos de activación 'relu' y tanh', pero se terminó optando por relu dado que terminó dando mejores resultados. El batch Size utilizado para entrenar fue de 128. La función de loss utilizada es la misma que utilizada en el ejemplo de Deepmind, la función Huber. Se encontró que la función Huber al ser una combinación del MSE para valores chicos y MAE para valores grandes, resulta en una buena relación de compromiso para tipos de entrenamiento reforzado.

En relación al optimizador utilizado en el paper de Deepmind se utilizó RMSProp, pero se encontró que utilizando Adam se obtienen mejores tiempos de entrenamiento.

La actualización de los parámetros de la red se da siguiendo la ecuación de las 'temporal differences'. En función de los estados y acciones tomadas se obtiene una predicción de los Q-values asociados (utilizando el modelo target para estabilidad). Para este cálculo se probó utilizando distintos 'discount factors', rondando sus valores principales entre 0.9 y 0.99.

Habiendo obtenido el valor de reward esperado futuro se obtuvo el Q value para las acciones dadas con el modelo no actualizado de la red. Teniendo ambas predicciones se calcula la loss y se realiza la propagación apropiada.

Ahora respecto al entrenamiento de la red en el contexto de Q learning se decidió definir un parámetro 'epsilon' que tome valores entre 0 y 1 que defina la proporción de exploración vs explotación. Se decidió ir variando este parámetro a lo largo de los episodios de entrenamiento en base a una función logística, permitiendo que al comienzo exista un proceso principalmente explorativo y luego se priorice la explotación.

Viendo los ejemplos encontrados en Internet se encontró que para estabilizar la obtención de los valores Q se utilizan 2 redes en paralelo corriendo el mismo modelo, una red base y una target. La red target se actualiza cada una cierta cantidad de acciones realizadas y es la que se ocupa de realizar las predicciones de los valores Q para la ecuación de las temporal Differences. Mientras tanto la red base es la que es entrenada acción a acción.
